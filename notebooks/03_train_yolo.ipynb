```python
from ultralytics import YOLO
from pathlib import Path

model = YOLO('../runs/detect/pcb_defect_detector_v2/weights/best.pt')

BASE_DIR = Path.cwd()
RUNS_DIR = BASE_DIR.parent / "runs" / "detect"
DATA_CONFIG = BASE_DIR.parent / "datasets" / "data.yaml"

model.train(
    data=str(DATA_CONFIG),
    project=str(RUNS_DIR),
    epochs=30,
    imgsz=1024,
    batch=4,
    lr0=0.0001,          # Уменьшенный шаг обучения
    close_mosaic=10,     # Выключить мозаику за 10 эпох до конца
    name='pcb_defect_detector_v3'
)
```
