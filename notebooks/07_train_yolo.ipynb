```python
from ultralytics import YOLO
from pathlib import Path

BASE_DIR = Path.cwd()
PRETRAINED_DIR = BASE_DIR.parent / "models" / "pretrained"
RUNS_DIR = BASE_DIR.parent / "runs" / "detect"
DATA_CONFIG = BASE_DIR.parent / "datasets" / "pcb-defects-1-update-cvat" / "data.yaml"

model = YOLO(str(PRETRAINED_DIR / 'yolo11l.pt'))

model.train(
    data=str(DATA_CONFIG),
    project=str(RUNS_DIR),
    epochs=50,
    imgsz=640,
    batch=8,
    amp=True,
    cache=True,
    workers=8,
    optimizer='AdamW',
    lr0=0.001,
    lrf=0.01,
    patience=10,
    close_mosaic=10,
    degrees=180.0,
    scale=0.5,
    flipud=0.5,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.1,
    copy_paste=0.2,
    name='pcb_defect_detector_v7'
) 
```

```python
from ultralytics import YOLO
from pathlib import Path

BASE_DIR = Path.cwd()
PRETRAINED_DIR = BASE_DIR.parent / "models" / "pretrained"
RUNS_DIR = BASE_DIR.parent / "runs" / "detect"
DATA_CONFIG = BASE_DIR.parent / "datasets" / "pcb-defects-final" / "data.yaml"

model = YOLO(str(PRETRAINED_DIR / 'yolo11l.pt'))

model.train(
    data=str(DATA_CONFIG),
    project=str(RUNS_DIR),
    epochs=150,
    imgsz=1024,
    batch=4,
    cache=True,
    workers=4,

    # Оптимизация обучения
    lr0=0.0001,
    patience=30,

    # Аугментации
    mosaic=0.5,
    mixup=0.1,
    copy_paste=0.3,

    # Важные параметры для финала
    close_mosaic=20,
    label_smoothing=0.1,
    cls=1.5,

    name='pcb_defect_detector_v12'
)
```
