```python
from ultralytics import YOLO
from pathlib import Path

BASE_DIR = Path.cwd()
PRETRAINED_DIR = BASE_DIR.parent / "runs" / "detect" / "pcb_defect_detector_v5" / "weights" / "best.pt"
RUNS_DIR = BASE_DIR.parent / "runs" / "detect"
DATA_CONFIG = BASE_DIR.parent / "datasets" / "pcb-defects-1-update-cvat" / "data.yaml"

model = YOLO(str(PRETRAINED_DIR))

model.train(
    data=str(DATA_CONFIG),
    project=str(RUNS_DIR),
    epochs=20,
    imgsz=1024,
    batch=4,
    overlap_mask=True, 
    close_mosaic=10,
    iou=0.5,
    save_period=5,
    name='pcb_defect_detector_v6'
)
```
