```python
from ultralytics import YOLO
from pathlib import Path

BASE_DIR = Path.cwd()
PRETRAINED_DIR = BASE_DIR.parent / "models" / "pretrained"
RUNS_DIR = BASE_DIR.parent / "runs" / "detect"
DATA_CONFIG = BASE_DIR.parent / "datasets" / "pcb-defects-1-update-cvat-v2-small-v2" / "data.yaml"

model = YOLO(str(PRETRAINED_DIR / 'yolo11l.pt'))

model.train(
    data=str(DATA_CONFIG),
    project=str(RUNS_DIR),
    epochs=100,
    imgsz=1024,
    batch=4,
    cache=True,
    workers=2,
    lr0=0.0001,
    patience=25,
    mosaic=0.5,
    mixup=0.1,
    copy_paste=0.3,
    close_mosaic=15,
    name='pcb_defect_detector_v9_small_tuned'
)
```
