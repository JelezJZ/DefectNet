```python
from ultralytics import YOLO

model = YOLO('yolo11n.pt')

# model = YOLO('runs/detect/train5/weights/last.pt')

# results = model.train(resume=True)

results = model.train(
    data='../data/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='pcb_defect_detector'
)

# Валидация
metrics = model.val()
print(f"mAP50: {metrics.box.map50}")
print(f"mAP50-95: {metrics.box.map}")

# Инференс на тестовом изображении
results = model.predict('../test_image.jpg', save=True)
```
